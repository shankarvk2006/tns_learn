# Pandas (DataFrames & Data Manipulation)

**Pandas** is a popular Python library used for data analysis and manipulation. It introduces two main data structures: **Series** (1D column) and **DataFrame** (2D table with rows and columns, similar to an Excel sheet).

### Key Concepts:

| Concept | Description |
| :--- | :--- |
| **Series** | A single column of data |
| **DataFrame** | A multi-column table (made of rows and columns) |
| **Filtering** | Selecting rows based on specific condition logic |
| **Grouping** | Aggregating data based on unique categories |
---

### Core Functions Reference:

| Function | Usage |
| :--- | :--- |
| `pd.Series()` | Create a 1D labeled array |
| `pd.DataFrame()` | Create a 2D labeled tabular structure |
| `pd.read_csv()` / `pd.read_excel()` | Load data from CSV or Excel files |
| `df.head()` / `df.tail()` | View first or last $n$ rows |
| `df.info()` / `df.describe()` | Get structural summary and statistical measures |
| `df['col']` / `df[['col1','col2']]` | Access single or multiple columns |
| `df.loc[]` / `df.iloc[]` | Access rows/columns by label (`loc`) or integer index (`iloc`) |
| `df.isnull()` / `df.dropna()` / `df.fillna()` | Detect, remove, or replace missing values |
| `df.drop(columns=[])` | Drop specified columns |
| `df.sort_values()` | Sort rows by values in a column |
| `df.groupby()` | Group data for aggregation |
| `df.merge()` / `df.join()` / `pd.concat()` | Combine multiple DataFrames |
| `df.to_csv()` | Save DataFrame to a CSV file |





1. Creation & Loading Data

```python
import pandas as pd

# Create Series and DataFrame
s = pd.Series([10, 20, 30], name="Numbers")
df = pd.DataFrame({"Name": ["Divya", "Arjun"], "Score": [85, 90]})

print("Series:\n", s)
print("\nDataFrame:\n", df)

# Load data from external files (syntax reference)
# df_csv = pd.read_csv("data.csv")
# df_excel = pd.read_excel("data.xlsx")
```

2. Viewing & Selecting Data (loc / iloc)

```python
import pandas as pd

data = {"Name": ["Divya", "Arjun", "Eshwar"], "Age": [22, 25, 21], "City": ["Chennai", "Bangalore", "Chennai"]}
df = pd.DataFrame(data)

# Access single vs multiple columns
print("Single Column:\n", df["Name"])
print("\nMultiple Columns:\n", df[["Name", "City"]])

# Access using iloc (index position)
print("\nFirst row using iloc:\n", df.iloc[0])

# Access using loc (label-based)
print("\nName and City of row 0 using loc:\n", df.loc[0, ["Name", "City"]])
```

3. Handling Missing Values & Dropping Columns

```python
import pandas as pd
import numpy as np

# DataFrame with missing values (NaN)
data = {"Name": ["Divya", "Arjun", "Eshwar"], "Score": [85, np.nan, 90], "Extra": [1, 2, 3]}
df = pd.DataFrame(data)

# Detect missing values
print("Check Nulls:\n", df.isnull())

# Fill missing values with 0
df["Score"] = df["Score"].fillna(0)

# Drop an unwanted column
df = df.drop(columns=["Extra"])
print("\nCleaned DataFrame:\n", df)
```

4. Sorting, Grouping, Combining & Saving

```python
import pandas as pd

df1 = pd.DataFrame({"ID": [1, 2], "Name": ["Divya", "Arjun"], "Score": [85, 90]})
df2 = pd.DataFrame({"ID": [1, 2], "City": ["Chennai", "Bangalore"]})

# Sort by Score descending
df1_sorted = df1.sort_values(by="Score", ascending=False)
print("Sorted:\n", df1_sorted)

# Merge two DataFrames on 'ID'
merged_df = pd.merge(df1, df2, on="ID")
print("\nMerged:\n", merged_df)

# Groupby example
print("\nAverage Score:\n", merged_df.groupby("City")["Score"].mean())

# Save DataFrame to CSV
# merged_df.to_csv("output.csv", index=False)
```
