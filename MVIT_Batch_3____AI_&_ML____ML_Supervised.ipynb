# Supervised Learning — Core Concepts

**Supervised Learning** is like teaching a student with an answer key: you feed the computer past questions (inputs) along with the correct answers (labels), so it learns how to predict answers for new questions.

---

### Core Concepts Reference:

| Algorithm / Concept | Simple Analogy | Goal |
| :--- | :--- | :--- |
| **Simple Linear Regression** | Drawing a straight trendline on a graph | Predict a number using **1** feature (e.g., predict marks based on study hours) |
| **Multiple Linear Regression** | Drawing a trendline using multiple factors | Predict a number using **2+** features (e.g., predict marks based on study hours AND attendance) |
| **Logistic Regression** | A Yes/No decision threshold | Predict a binary group (e.g., Pass vs. Fail) |
| **Decision Tree** | A flow-chart of simple IF-ELSE questions | Make decisions by splitting data into branches |
| **Random Forest** | Asking a team of experts instead of just one | Combines predictions from multiple Decision Trees for better accuracy |
| **Evaluation Metrics** | A scorecard for your AI model | Measures how accurately the model made predictions |

---

#### 1. Linear Regression (Predicting Marks)

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Step 1: Create simple student data
data = {
    "Hours": [2, 4, 6, 8, 10],
    "Attendance": [60, 70, 80, 90, 95],
    "Score": [40, 55, 70, 85, 95] # Target number to predict
}
df = pd.DataFrame(data)

# --- A. SIMPLE LINEAR REGRESSION (1 Input Feature) ---
X_simple = df[["Hours"]]  # Input: Only Hours
y = df["Score"]           # Output: Marks

simple_model = LinearRegression()
simple_model.fit(X_simple, y) # Train the model

# Predict score for a student studying 5 hours
pred1 = simple_model.predict([[5]])
print("Simple Regression (5 hrs):", pred1[0])


# --- B. MULTIPLE LINEAR REGRESSION (2 Input Features) ---
X_multi = df[["Hours", "Attendance"]] # Inputs: Hours AND Attendance

multi_model = LinearRegression()
multi_model.fit(X_multi, y) # Train the model

# Predict score for 5 hours study AND 75% attendance
pred2 = multi_model.predict([[5, 75]])
print("Multiple Regression (5 hrs, 75% attendance):", pred2[0])
```

### 2. Logistic Regression & Model Evaluation Scorecard

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Step 1: Create dataset (0 = Fail, 1 = Pass)
data = {
    "Hours": [1, 2, 3, 4, 5, 6, 7, 8],
    "Passed": [0, 0, 0, 0, 1, 1, 1, 1]
}
df = pd.DataFrame(data)

X = df[["Hours"]]
y = df["Passed"]

# Step 2: Split data (Train the model on 75% data, test on 25% data)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# Step 3: Train Logistic Regression Model
model = LogisticRegression()
model.fit(X_train, y_train)

# Step 4: Ask the model to guess test results
y_pred = model.predict(X_test)

# --- SCORECARD METRICS ---
# Accuracy: Percentage of total correct predictions
print("Accuracy: ", accuracy_score(y_test, y_pred))

# Precision: Out of all predicted PASSES, how many were actual PASSES?
print("Precision:", precision_score(y_test, y_pred))

# Recall: Out of all actual PASSES, how many did the model catch?
print("Recall:   ", recall_score(y_test, y_pred))

# F1-Score: The overall balance between Precision and Recall
print("F1-Score: ", f1_score(y_test, y_pred))
```

### 3. Decision Trees vs. Random Forests

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

# Step 1: Create student dataset
data = {
    "Hours": [2, 3, 5, 7, 8, 10],
    "Attendance": [50, 60, 75, 85, 90, 95],
    "Passed": [0, 0, 1, 1, 1, 1] # 0 = Fail, 1 = Pass
}
df = pd.DataFrame(data)

X = df[["Hours", "Attendance"]]
y = df["Passed"]

# --- 1. DECISION TREE (Single Flow-Chart Tree) ---
dt_model = DecisionTreeClassifier()
dt_model.fit(X, y)

# Predict for student with 6 hours study and 80% attendance
dt_result = dt_model.predict([[6, 80]])
print("Decision Tree says:", "PASS" if dt_result[0] == 1 else "FAIL")


# --- 2. RANDOM FOREST (Team of 10 Decision Trees Voting Together) ---
rf_model = RandomForestClassifier(n_estimators=10, random_state=42)
rf_model.fit(X, y)

# Predict for the same student
rf_result = rf_model.predict([[6, 80]])
print("Random Forest says:", "PASS" if rf_result[0] == 1 else "FAIL")
```
