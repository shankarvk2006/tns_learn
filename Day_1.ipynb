```python
# def is a keyword used to define a function.

# 1. Defining a function named 'first' using 'def'
def first(numbers_list):
    """
    A simple function that receives a list
    and returns its very first element.
    """
    if len(numbers_list) == 0:
        return "List is empty!"

    return numbers_list[0]
```

```python
# 2. Calling/using the 'first' function

my_numbers = [10, 20, 30, 40]
result = first(my_numbers)

print("The first element is:", result)  # Output: 10
```

**ERROR AND DEBUGGING**

```python
# ==========================================
# 1. HANDLING COMMON ERRORS (try-except-else-finally)
# ==========================================
# Use try-except blocks to prevent your code from crashing when an error occurs.

def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError as e:
        print(f"[ERROR CATCH]: Cannot divide by zero! Details: {e}")
        return None
    except TypeError as e:
        print(f"[ERROR CATCH]: Invalid input type! Details: {e}")
        return None
    else:
        # Runs ONLY if no exceptions were raised
        print("[SUCCESS]: Division executed smoothly.")
        return result
    finally:
        # ALWAYS runs, regardless of whether an error occurred (useful for cleanup)
        print("[CLEANUP]: Division attempt finished.\n")

print("--- 1. Exception Handling ---")
safe_divide(10, 2)   # Success case
safe_divide(10, 0)   # Triggers ZeroDivisionError
safe_divide(10, "2") # Triggers TypeError
```

```python
# ==========================================
# 2. RAISING CUSTOM EXCEPTIONS
# ==========================================
# You can force an error using the `raise` keyword when specific business logic fails.

class InvalidAgeError(Exception):
    """Custom Exception for invalid age inputs."""
    pass

def verify_voter_age(age):
    if age < 0:
        raise InvalidAgeError("Age cannot be negative!")
    elif age < 18:
        raise InvalidAgeError("User must be at least 18 years old to vote.")
    else:
        print("Access granted: Eligible to vote!")

print("--- 2. Custom Errors ---")
try:
    verify_voter_age(15)
except InvalidAgeError as e:
    print(f"[CUSTOM ERROR CATCH]: {e}\n")
```
