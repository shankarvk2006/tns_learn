# Python Binary Trees

A **Tree** is a hierarchical data structure. A **Binary Tree** is a tree where each node has at most **two children** (referred to as the `left` child and `right` child).

### Terminology:

* **Root:** The topmost node of the tree.
* **Parent / Child:** A node pointing down to another node is the parent; the node below is the child.
* **Leaf:** A node with no children (`left` and `right` are both `None`).

---

### Core Structure:

| Component | Description |
| :--- | :--- |
| `Node` | Holds `data`, a pointer to `left` child, and a pointer to `right` child |
| **Inorder Traversal** | Visit nodes in Left $\rightarrow$ Root $\rightarrow$ Right order |

```python
# Step 1: Create a Tree Node
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

# Step 2: Build a simple Binary Tree manually
root = Node("A")
root.left = Node("B")
root.right = Node("C")

# Visual Representation:
#       A
#      / \
#     B   C

# Step 3: Print tree nodes (Inorder Traversal)
def print_tree(node):
    if node:
        print_tree(node.left)   # Visit left
        print(node.data, end=" ") # Visit root
        print_tree(node.right)  # Visit right

print_tree(root)
# Output: B A C
```

# Python Graphs

A **Graph** is a non-linear data structure made up of **Nodes** (also called Vertices) connected by **Edges** (lines showing relationships).

### Terminology:

* **Vertex (Node):** An individual data element (e.g., a person or a city).
* **Edge:** The link or connection between two nodes.
* **Adjacency List:** A popular way to represent a graph using a Python dictionary, where keys are nodes and values are lists of connected nodes.

---

### Representation:

| Method | Description |
| :--- | :--- |
| **Adjacency List** | Stores each node and its connected neighbours using a `dict` |

---

```python
# Step 1: Represent a Graph using a Dictionary (Adjacency List)
graph = {
    "A": ["B", "C"],  # Node A is connected to B and C
    "B": ["A", "D"],  # Node B is connected to A and D
    "C": ["A"],       # Node C is connected to A
    "D": ["B"]        # Node D is connected to B
}

# Visual Representation:
#    A --- C
#    |
#    B --- D

# Step 2: Print all connections
for node, neighbours in graph.items():
    print(f"Node {node} is connected to: {neighbours}")

# Output:
# Node A is connected to: ['B', 'C']
# Node B is connected to: ['A', 'D']
# Node C is connected to: ['A']
# Node D is connected to: ['B']
```

# Python Hash Tables

A **Hash Table** is a data structure that stores key-value pairs. In Python, Hash Tables are implemented built-in as **Dictionaries (`dict`)**.

It uses a **Hash Function** to turn a key into an index, allowing super-fast $O(1)$ lookup time.

### Terminology:

* **Key:** The unique identifier used to look up a value.
* **Value:** The data stored under a specific key.
* **Hash Function:** Converts a key into a fixed numerical index in memory.
* **Collision:** When two different keys hash to the same position (handled automatically by Python).

---

### Key Concepts:

| Concept | Description |
| :--- | :--- |
| **Direct Lookup** | Access data instantly using key instead of searching sequentially ($O(1)$) |
| **Key Uniqueness** | Keys must be unique; assigning a value to an existing key updates it |

---

```python
# Step 1: Create a Hash Table (Dictionary)
hash_table = {}

# Step 2: Insert Key-Value Pairs
hash_table["name"] = "Divya"
hash_table["role"] = "Trainer"
hash_table["id"] = 101

# Step 3: Fast Lookup by Key
print("Name:", hash_table["name"])  # Output: Divya
print("Role:", hash_table["role"])  # Output: Trainer

# Step 4: Check if a key exists
if "id" in hash_table:
    print("ID found:", hash_table["id"]) # Output: ID found: 101
```
